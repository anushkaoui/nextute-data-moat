# Early model on synthetic data

**Case B · The Data Moat · Nextute × Finance Club IIT Patna 2026**

This is the optional Round 1 model artefact. It demonstrates:

1. **Rasch/1PL recovery** of student ability and item difficulty.
2. A **time-like dropout split**: train on the earliest cohort, calibrate on the middle cohort, test on the latest cohort.
3. **Calibration and top-decile precision**, not just AUC.

> **Caveat:** everything here is synthetic. No real student data is used. The dropout labels contain a designed signal and measurement noise. These results validate the pipeline and evaluation method, not real-world effectiveness.

## Inspect the saved results

The script writes a JSON file so the metrics can be cited without re-running the notebook.


```python
import json
from pathlib import Path

results = json.loads(Path('../docs/early_model_results.json').read_text())
print('Seed:', results['seed'])
print('Rasch theta correlation:', results['rasch']['theta_correlation'])
print('Rasch b correlation:', results['rasch']['b_correlation'])
print('Dropout AUC:', results['dropout']['auc'])
print('Dropout PR-AUC:', results['dropout']['pr_auc'])
print('Dropout Brier:', results['dropout']['brier'])
print('Dropout precision@10%:', results['dropout']['precision_at_10pct'])
print('Test base rate:', results['dropout']['test_base_rate'])
```

## Interpretation

- **Rasch recovery:** ability correlation ≈ 0.92 and difficulty correlation ≈ 0.99 on 42,000 synthetic responses. This shows the estimator recovers the designed latent structure; it is not evidence about real institutes.
- **Dropout baseline:** AUC ≈ 0.74, PR-AUC ≈ 0.21 against a ~9% base rate, Brier ≈ 0.078. This is deliberately not an AUC of 1.0. It shows a realistic, imperfect baseline.
- **Calibration:** predicted probabilities are close to observed rates in the available bins. The test cohort has few high-risk cases, so the top of the calibration curve is sparse.
- **What is missing:** public-dataset benchmarking (EdNet/ASSISTments/OULAD) and real, consented, time-based validation. Both are required before making real-world claims.
